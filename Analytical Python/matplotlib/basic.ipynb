# Matplotlib

```python
import matplotlib.pyplot as plt
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [12000, 13500, 12800, 14500, 16000, 15500]
```

```python
# Plotting line graph for monthly revenue growth plot(x, y)
plt.plot(months, revenue)
# Adding the title of the graph
plt.title("Monthly Revenue Growth")
# Adding label to the x-axis
plt.xlabel("Months")
# Adding label to the y-axis
plt.ylabel("Revenue (in rupees)")
# Printing the graph
plt.show()
```

```python
# Product sales
products = ["Apples", "Bananas", "Oranges", "Grapes", "Mangoes"]
sales = [150, 200, 180, 120, 210]
```

```python
# Plotting bar graph for sales of the products - bar(x, y)
plt.bar(products, sales)
# Adding the title of the graph
plt.title("Sales of the products")
# Adding label to the x-axis
plt.xlabel("Products")
# Adding label to the y-axis
plt.ylabel("Sales (quantity)")
# Showing the bar graph
plt.show()
```

```python
# Average marks of subjects
subjects = ["Math", "Science", "English", "History", "Computer"]
avg_marks = [78, 85, 74, 80, 90]
```

```python
# For colouring
colours = ["lightblue", "lightgreen", "peach", "gray", "lightyellow"]
# Plotting pie-chart for sales of the products - piw(x, y)
plt.pie(avg_marks, labels=subjects, autopct="%1.01f%%", colors=colours)
# Adding the title of the pie-chart
plt.title("Average marks of subjects")
plt.show()
```
