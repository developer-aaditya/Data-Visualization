```python
# import pandas library
import pandas as pd
# Marks for the student for particular subject
marks = [60, 70, 80, 45, 90]
# Subject of the corresponding marks
subjects = ["Hindi", "English", "Maths", "Computer", "Science"]
# Using pandas, transform data into the series
data_series = pd.Series(marks, index=subjects, name="Subject Marks")
# print the data series
print(data_series)
```

```python
# DataFrame in the pandas
```

```python
data = {'Id':['std101', 'std102', 'std103', 'std104'],
        'Name': ['Anita', 'Rahul', 'Monu', 'Sahil'],
        'Rollno': [21, 32, 24, 42],
        'Standard': ['10th', '6th', '9th', '11th']}
# to create a dataframe
df = pd.DataFrame(data)
# To insert a new coloumn
df['Hindi'] = [70, 77, 89, 58]
print(df)
```

```python
# creating a new record(row)
record1 = pd.DataFrame([{'Id':'std105',
            'Name':'Rakesh',
            'Rollno':23,
            'Standard':'7th',
            'Hindi':89}])
df = pd.concat([df, record1], ignore_index=True)
print(df)
```
